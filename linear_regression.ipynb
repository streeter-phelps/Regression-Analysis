# Multiple Regressions in a Single Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import linregress

# Sample data with different variables and colors
data = {
    'x1': [1, 2, 3, 4, 5],
    'y1': [2, 3, 5, 4, 5],
    'x2': [1, 2, 3, 4, 5],
    'y2': [4, 3, 2, 4, 5],
    'x3': [1, 2, 3, 4, 5],
    'y3': [1, 3, 4, 2, 5]
}

# Create a DataFrame from the data dictionary
df = pd.DataFrame(data)

# Use regplot with hue to show different colors for different variables
sns.regplot(x='x1', y='y1', data=df, color='red', label='Variable 1')
sns.regplot(x='x2', y='y2', data=df, color='green', label='Variable 2')
sns.regplot(x='x3', y='y3', data=df, color='blue', label='Variable 3')

# Calculate linear regression statistics for each variable (you can customize this part)
slope1, intercept1, r_value1, p_value1, std_err1 = linregress(df['x1'], df['y1'])
slope2, intercept2, r_value2, p_value2, std_err2 = linregress(df['x2'], df['y2'])
slope3, intercept3, r_value3, p_value3, std_err3 = linregress(df['x3'], df['y3'])

# Annotate the plot with the correlation coefficients (r) and p-values for each variable
plt.annotate(f"Var 1: r={r_value1:.2f}, p={p_value1:.4f}\nVar 2: r={r_value2:.2f}, p={p_value2:.4f}\nVar 3: r={r_value3:.2f}, p={p_value3:.4f}",
             xy=(0.05, 0.85), xycoords='axes fraction', fontsize=10, color='black')

# Add legend
plt.legend()

# Save the plot as a PNG file
plt.savefig('regression_plot_multi_variable.png')

# Show the plot
plt.show()
```
